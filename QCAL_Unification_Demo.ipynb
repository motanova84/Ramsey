## Universal Constants

Display the fundamental constants that unify all problems.

````python
# Display constants as formatted table
print("📊 UNIVERSAL CONSTANTS")
print("=" * 70)

constants_data = []
for name, value in framework.constants.items():
    constants_data.append([name, f"{value:.6f}" if isinstance(value, float) else str(value)])

# Create markdown table
table_md = "| Constant | Value |\n|----------|-------|\n"
for row in constants_data:
    table_md += f"| {row[0]} | {row[1]} |\n"

display(Markdown(table_md))
````

## Problem Selector

Select a millennium problem to explore its QCAL connection.

````python
def show_qcal_connection(problem_key: str):
    """
    Display how a specific problem connects through QCAL.
    """
    if problem_key not in framework.problems:
        print(f"❌ Unknown problem: {problem_key}")
        return
    
    problem = framework.problems[problem_key]
    constant_key = problem['constant']
    constant_value = framework.constants[constant_key]
    
    # Get default parameters and compute eigenvalue
    params = framework._get_default_params(problem_key)
    operator = framework.operators[problem_key]
    eigenvalue = operator(params)
    
    # Get connections
    connections = framework.find_connections(problem_key)
    verification = framework.verify_problem(problem_key)
    
    # Display formatted information
    output = f"""
## {problem['name']} in QCAL Unified Framework

**Problem Statement**: {problem['statement']}

**QCAL Operator**: `{problem['operator']}`

**Universal Constant**: `{constant_key} = {constant_value}`

**Eigenvalue**: `{eigenvalue}`

**Verification Protocol**: {problem['verification']}

**Verification Status**: {verification}

**Connected Problems**: {', '.join([framework.problems[p]['name'] for p in connections]) if connections else 'None'}

### Connection Diagram

```
{problem['name']}
    ↓ (via {constant_key})
  QCAL Operator: {problem['operator']}
    ↓
  Eigenvalue: {eigenvalue}
```
    """
    
    display(Markdown(output))

# Example: Show P vs NP connection
show_qcal_connection('p_vs_np')
````

## Connection Graph Visualization

Visualize how all problems connect through QCAL.

````python
# Get all connections
connections = framework.get_all_connections()

# Create connection matrix for visualization
problems = list(framework.problems.keys())
n = len(problems)
connection_matrix = np.zeros((n, n))

for i, prob_i in enumerate(problems):
    for j, prob_j in enumerate(problems):
        if prob_j in connections[prob_i]:
            connection_matrix[i, j] = 1

# Plot connection matrix
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(connection_matrix, cmap='Blues', aspect='auto')

# Set ticks and labels
ax.set_xticks(np.arange(n))
ax.set_yticks(np.arange(n))
ax.set_xticklabels([framework.problems[p]['name'] for p in problems], rotation=45, ha='right')
ax.set_yticklabels([framework.problems[p]['name'] for p in problems])

# Add text annotations
for i in range(n):
    for j in range(n):
        if connection_matrix[i, j] > 0:
            text = ax.text(j, i, '✓', ha='center', va='center', color='white', fontsize=16)

ax.set_title('QCAL Problem Connection Matrix', fontsize=14, fontweight='bold')
plt.colorbar(im, ax=ax, label='Connected')
plt.tight_layout()
plt.show()

print("\n📊 Connection Summary:")
total_connections = int(np.sum(connection_matrix))
print(f"  Total connections: {total_connections}")
print(f"  Average connections per problem: {total_connections/n:.2f}")
````
