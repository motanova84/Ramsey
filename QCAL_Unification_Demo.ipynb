## Problem Connections

Explore how each millennium problem connects to others through QCAL:

````python
class QCALUnificationDemo:
    """Interactive dashboard for QCAL connections."""
    
    def __init__(self, framework):
        self.framework = framework
        self.problems = [
            "P vs NP", "Riemann Hypothesis", "BSD Conjecture",
            "Navier-Stokes", "Ramsey Numbers", "Yang-Mills", "Hodge Conjecture"
        ]
        self.problem_map = {
            "P vs NP": "p_vs_np",
            "Riemann Hypothesis": "riemann",
            "BSD Conjecture": "bsd",
            "Navier-Stokes": "navier_stokes",
            "Ramsey Numbers": "ramsey",
            "Yang-Mills": "yang_mills",
            "Hodge Conjecture": "hodge"
        }
    
    def create_dashboard(self):
        """Create interactive dashboard."""
        problem_selector = widgets.Dropdown(
            options=self.problems,
            description='Problem:',
            style={'description_width': 'initial'}
        )
        
        visualize_btn = widgets.Button(
            description="Show QCAL Connection",
            button_style='info'
        )
        
        output = widgets.Output()
        
        def on_visualize(b):
            with output:
                output.clear_output()
                problem = problem_selector.value
                self.show_qcal_connection(problem)
        
        visualize_btn.on_click(on_visualize)
        
        display(widgets.VBox([
            widgets.HTML("<h3>QCAL Problem Explorer</h3>"),
            problem_selector,
            visualize_btn,
            output
        ]))
    
    def show_qcal_connection(self, problem_name):
        """Display how problem connects through QCAL."""
        problem_key = self.problem_map[problem_name]
        connections = self.get_qcal_connections(problem_key)
        
        md_output = f"""
## {problem_name} in QCAL Unified Framework

**QCAL Operator**: `{connections['operator']}`

**Universal Constant**: `{connections['constant']}`

**Problem Statement**: {connections['statement']}

**Connection Equation**:
```
{connections['equation']}
```

**Verification Protocol**: {connections['verification']}

**Connected Problems**: {', '.join(connections['connected_to'])}

**Verification Status**: {connections['status']}
        """
        
        display(Markdown(md_output))
    
    def get_qcal_connections(self, problem_key):
        """Get connection details for a problem."""
        metadata = self.framework.problem_metadata.get(problem_key, {})
        
        # Get connections
        connected_to = self.framework._find_connections(problem_key)
        connected_names = [self.framework.problem_metadata[k]['name'] 
                          for k in connected_to]
        
        # Get verification status
        status = self.framework._verify_problem(problem_key)
        
        # Create equation based on problem
        equations = {
            'p_vs_np': 'D_PNP(φ) = κ_Π · log(tw(G_I(φ))) ≥ κ_Π · tw(φ)/log(n)',
            'riemann': 'H_Ψ(z) = 0 ↔ Re(z) = 1/2, Im(z) = 2πf₀',
            'bsd': 'L_E(1) = Δ · Ω_E · Reg_E · ∏_p c_p / |E_tors|²',
            'navier_stokes': '∂u/∂t + (u·∇)u = -∇p + ν∇²u + ε·R_QCAL[u]',
            'ramsey': 'R_ψ(r,s) = O(√(rs) · ln(rs)) via f₀ = 141.7 Hz',
            'yang_mills': 'Mass gap = g_YM · f₀ / 100 GeV',
            'hodge': 'h^{1,1} + h^{2,1} = 13 for quintic Calabi-Yau'
        }
        
        return {
            'operator': metadata.get('operator', ''),
            'constant': metadata.get('constant', ''),
            'statement': metadata.get('statement', ''),
            'verification': metadata.get('verification', ''),
            'connected_to': connected_names,
            'status': status,
            'equation': equations.get(problem_key, 'N/A')
        }

# Create and display dashboard
demo = QCALUnificationDemo(framework)
demo.create_dashboard()
````

## Ramsey Number Demonstration

Show the vibrational reduction of Ramsey numbers:

````python
# Interactive Ramsey calculator
r_slider = widgets.IntSlider(value=5, min=3, max=8, description='r:')
s_slider = widgets.IntSlider(value=5, min=3, max=8, description='s:')
ramsey_output = widgets.Output()

def calculate_ramsey(change):
    with ramsey_output:
        ramsey_output.clear_output()
        r = r_slider.value
        s = s_slider.value
        
        # Calculate R_ψ
        r_psi = framework.R_operator({'r': r, 's': s})
        
        # Classical bound (simplified)
        import math
        classical_bound = int(2 ** (math.sqrt(r + s) * math.log(r + s)))
        
        print(f"R_ψ({r},{s}) = {r_psi}")
        print(f"Classical bound ≈ 2^(√{r+s} · ln({r+s})) ≈ {classical_bound}")
        print(f"Reduction factor: {classical_bound / max(r_psi, 1):.2f}x")
        print(f"\nVibrational frequency: {framework.constants['f0']} Hz")
        print(f"Ramsey ratio φ_R: {framework.constants['ramsey_ratio']:.6f}")

r_slider.observe(calculate_ramsey, 'value')
s_slider.observe(calculate_ramsey, 'value')

display(widgets.VBox([
    widgets.HTML("<h3>Vibrational Ramsey Calculator</h3>"),
    r_slider,
    s_slider,
    ramsey_output
]))

# Trigger initial calculation
calculate_ramsey(None)
````
